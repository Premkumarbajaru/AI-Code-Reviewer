```python
import os
from langchain_groq import ChatGroq
```

```python
os.environ["GROQ_API_KEY"] = "enter groq api key"
```

```python
llm = ChatGroq(
    model ="llama-3.1-8b-instant",
    temperature = 0
)
```

```python
response = llm.invoke("Write me a song about sunrise in 2000words")
print(response.content)
```

### optimizied token by token

```python
for chunk in llm.stream("Write me a song about sunrise in 2000words"):
  print(chunk.content, end="", flush=True)
```
